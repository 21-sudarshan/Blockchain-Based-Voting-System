```python
# Blockchain-Based Voting System

import hashlib

# Store votes
votes = {
    "Candidate A": 0,
    "Candidate B": 0
}

# Store voter IDs to prevent duplicate voting
voters = []

# Voting Function
def vote(voter_id, candidate):

    if voter_id in voters:
        print(f"Voter {voter_id}: Already Voted")
        return

    if candidate not in votes:
        print("Invalid Candidate")
        return

    voters.append(voter_id)
    votes[candidate] += 1

    print(f"Voter {voter_id}: Vote Recorded for {candidate}")

# Cast Votes
vote("101", "Candidate A")
vote("102", "Candidate B")
vote("103", "Candidate A")
vote("101", "Candidate B")  # Duplicate vote test

# Display Results
print("\n===== VOTING RESULTS =====")
for candidate, count in votes.items():
    print(f"{candidate}: {count} votes")

# Create Blockchain Hash
data = str(votes)
block_hash = hashlib.sha256(data.encode()).hexdigest()

print("\n===== BLOCKCHAIN SECURITY =====")
print("Blockchain Hash:")
print(block_hash)

# Display Voters
print("\n===== REGISTERED VOTERS =====")
print(voters)
```

```python

# Save results to a text file

with open("VotingResults.txt", "w") as file:
    file.write("BLOCKCHAIN-BASED VOTING SYSTEM\n")
    file.write("=" * 40 + "\n\n")

    file.write("Voting Results:\n")
    for candidate, count in votes.items():
        file.write(f"{candidate}: {count} votes\n")

    file.write("\nBlockchain Hash:\n")
    file.write(hash_value)

print("Results saved to VotingResults.txt")
```
